Task 3: Random Agent

```python
import random

class VacuumEnvironment:

    def __init__(self):
        self.status = {
            'A': random.randint(0, 1),
            'B': random.randint(0, 1)
        }
        self.agent_location = random.choice(['A', 'B'])

    def percept(self):
        return (self.agent_location,
                self.status[self.agent_location])

    def execute(self, action):

        if action == 'Suck':
            self.status[self.agent_location] = 0

        elif action == 'Right':
            self.agent_location = 'B'

        elif action == 'Left':
            self.agent_location = 'A'


def random_agent(percept):

    return random.choice(['Left', 'Right', 'Suck'])


random.seed(7)

total_score = 0

for trial in range(1, 11):

    env = VacuumEnvironment()
    score = 0

    for step in range(20):

        p = env.percept()
        action = random_agent(p)

        env.execute(action)

        if action == 'Suck':
            score += 10
        else:
            score -= 1

    total_score += score

    print("Trial:", trial, "Score:", score)

average = total_score / 10

print("Average Performance:", average)
```

Task 4: Model-Based Agent


```python
class VacuumEnvironment:

    def __init__(self):
        self.status = {'A': 1, 'B': 1}
        self.agent_location = 'B'

    def percept(self):
        return (self.agent_location,
                self.status[self.agent_location])

    def execute(self, action):

        if action == 'Suck':
            self.status[self.agent_location] = 0

        elif action == 'Right':
            self.agent_location = 'B'

        elif action == 'Left':
            self.agent_location = 'A'


def model_based_agent(percept, memory):

    location, dirty = percept

    if dirty == 1:
        memory[location] = False
        return 'Suck'

    else:
        memory[location] = True

    if memory['A'] and memory['B']:
        return 'NoOp'

    if location == 'A':
        return 'Right'

    else:
        return 'Left'


# Create environment
env = VacuumEnvironment()

# Memory of cleaned locations
memory = {'A': False, 'B': False}

score = 0

print("Initial state:", env.status)
print("Agent location:", env.agent_location)

for step in range(1, 10):

    p = env.percept()

    action = model_based_agent(p, memory)

    if action == 'NoOp':
        print("Step", step, ": NoOp - Both locations are clean")
        break

    env.execute(action)

    if action == 'Suck':
        score += 10
    else:
        score -= 1

    print("Step", step, ":",
          "Percept =", p,
          "Action =", action,
          "World =", env.status)

print("Final state:", env.status)
print("Memory:", memory)
print("Performance Score:", score)
```
